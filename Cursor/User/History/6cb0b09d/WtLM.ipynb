```python
from gradio_client import Client

client_gr = Client("http://127.0.0.1:7860/")
```

```python
async def callback(message):
    name = message.get("name", None)
    messages = message.get("recent_message", None)
    time = message.get("time", None)
    unreads = message.get("unread_messages", None)
    print(name, messages, time, unreads)

    if int(unreads) > 0:
        result = client_gr.predict(
            user_input=str(messages),
            recipient=name,
            chat_history="",
            timestamp=time,
            api_name="/predict",
        )
        await client.send_message(name, result)
```

```python
messages = []
while True:
    new_message = await client.fetch_latest_message()
    if new_message not in messages:

        messages.append(new_message)
        await callback(new_message)
    else:
        continue

    await asyncio.sleep(1)
```

```python
iteration_count = 0
while True:
    if iteration_count % 30 == 0:
        print("Checking for scheduled messages")
    iteration_count += 1
    await asyncio.sleep(1)
```
