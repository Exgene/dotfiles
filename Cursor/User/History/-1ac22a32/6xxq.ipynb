```python
class GraphNode:
    def __init__(self, name: str, value=0):
        self.name = name
        self.value = value


def calculate_value_from_episodes(ep1, ep2, name):
    sumValue = 0
    i = 0

    while i < len(ep1) and ep1[i].name != name:
        i += 1

    if i < len(ep1):
        sumValue += ep1[i].value

    print(sumValue)
    i = 0

    while i < len(ep2) and ep2[i].name != name:
        i += 1

    if i < len(ep2):
        print(ep2[i].value)
        sumValue += ep2[i].value

    print(sumValue)

    return sumValue / 2
```

```python
episode_1 = [
    GraphNode("A", 3),
    GraphNode("A", 2),
    GraphNode("B", -4),
    GraphNode("A", 4),
    GraphNode("B", -3),
]
```

```python
episode_2 = [
    GraphNode("B", -2),
    GraphNode("A", 3),
    GraphNode("B", -3),
]
```

```python
calculate_value_from_episodes(episode_1, episode_2, "A")  # 2.5
```
