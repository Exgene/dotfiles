```python
class GraphNode:
    def __init__(self, name: str, value=0):
        self.name = name
        self.value = value

    def calculate_value_from_episodes(ep1, ep2, name):
        sumValue = 0
        i = 0

        while i < len(ep1) and ep1[i].name != name:
            i += 1

        if i < len(ep1):
            sumValue += ep1[i].value
        i = 0

        while i < len(ep2) and ep2[i].name != name:
            i += 1

        if i < len(ep2):
            sumValue += ep2[i].value

        return sumValue / 2
```
