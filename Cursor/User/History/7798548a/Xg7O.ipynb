```python
async def periodic_chat_update(client: WhatsappClient, interval: int = 20):
    """
    Periodically fetches and updates chats for specified users.

    :param client: WhatsappClient instance
    :param interval: Time interval in seconds (default: 20)
    """
    while True:
        await client.update_database("Prathwik")
        await asyncio.sleep(interval)
```

```python
await client.login()
```

```python
await periodic_chat_update(client)
```

```python
import sqlite3
from newFile import get_db_path
from prettytable import PrettyTable


def get_chats_from_db():
    """Gets all chats from the chats table."""
    db_path = get_db_path()
    connection = sqlite3.connect(db_path)
    cursor = connection.cursor()

    cursor.execute("SELECT * FROM chats")
    chats = cursor.fetchall()
    connection.close()
    return chats


# Fetch chats and display them in a table format
chats = get_chats_from_db()

# Create a PrettyTable instance
table = PrettyTable()
table.field_names = [
    "Chat ID",
    "Message ID",
    "Provider",
    "Sender",
    "Time Sent",
    "Message",
]

# Add rows to the table
for chat in chats:
    table.add_row(chat)

# Print the table
print(table)
```
